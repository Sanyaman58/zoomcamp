```python
print(df['horsepower'].median()) # median value
print(df['horsepower'].mode()[0]) # most frequent value
df['horsepower'] = df['horsepower'].fillna(df['horsepower'].mode()[0]) # filling the missing values int the horsepower column
print(df['horsepower'].median()) # median value decreased
```

```python
df_asia = df[df['origin']=="Asia"] # all Asia cars
X = df_asia[['model_year', 'vehicle_weight']].head(7).values # transforming Asian's model_year and vehicle_weight first 7 values into the array
XTX = X.T.dot(X) # getting the X.T @ X matrix-matrix multiplication
XTX_inv = np.linalg.inv(XTX) # getting the inverted XTX matrix
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
(XTX_inv.dot(X.T)).dot(y).sum() # getting the sum of all multiplied values
```
